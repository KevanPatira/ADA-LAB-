```python
def ReflexVacuumAgent(loc, statusA, statusB, cost):
  if loc == 'A':
    if statusA == 'D':
      print("Sweep at A")
      cost+=1
      statusA = 'C'
    if statusB == 'D':
      print("Move to right")
      cost+=1
      ReflexVacuumAgent('B', statusA, statusB,cost)
    else:
      print(f"Job Done. \nCost = {cost}")

  elif loc == 'B':
    if statusB == 'D':
      print("Sweep at B")
      cost+=1
      statusB = 'C'
    if statusA == 'D':
      print("Move to left")
      cost+=1
      ReflexVacuumAgent('A', statusA, statusB)

    else:
      print(f"Job Done. \nCost = {cost}")

  else:
    print("Enter valid Location")

loc = input("Enter Location: ").upper()
statusA = input("Enter Status of A: ").upper()
statusB = input("Enter Status of B: ").upper()
ReflexVacuumAgent(loc, statusA, statusB, 0)
```
